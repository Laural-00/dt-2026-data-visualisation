```python
Create a chart showing who puts in the most effort
```

```python
import pandas as pd
df = pd.read_csv("competencies.csv")
df
```

```python
import matplotlib.pyplot as plt
avg_hours = df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", legend=False, color="pink")
plt.ylabel("Average hours studied")
plt.title("Average hours studied per student")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*40, alpha=0.5, color="pink")
plt.xlabel("Hours studied")
plt.ylabel("Grades")
plt.title("Grades vs hours studied(Bubble size=hours)")
```

```python
plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Distribution of study hours and grades")
```
