```python
# Can use to work with tabular data
import pandas as pd 
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
df.describe() # Show some basic statistics
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Name", y="Years_of_practice")
plt.title("Years of practice per athlete")
plt.ylabel("Years of practice")
plt.show()
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age per athlete")
plt.ylabel("Age")
plt.show()
```

```python
df.plot(kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Age of athletes")
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60, alpha=0.6)
#s=df["Years_of_practice"]*60 this means if student has more years of practice, their bubble is bigger
#alpha=0.6 makes the bubbles a bit transparent
plt.xlabel("Age")
plt.ylabel("Years of practice")
plt.title("Age vs years of practice")
```

```python
import seaborn as sns #use this to create more clean
sns.boxplot(data=df, x="Age")
plt.title("Distribution of athletes' ages")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt.show()
```

```python
avg_age = df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar", legend=False)
plt.ylabel("Average age")
plt.title("Average age per sport")
plt.show()
```

```python
avg_years = df.groupby("Sport")["Years_of_practice"].mean()
avg_years.plot(kind="bar", legend=False)
plt.ylabel("Average years of practice")
plt.title("Average years of practice per sport")
plt.show()
```
