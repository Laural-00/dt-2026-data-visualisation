```python
import pandas as pd
df = pd.read_csv("China cities population.csv")
df
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Urban Area", y="Population Estimate (2025)", legend=False, color="pink")
plt.title("Population of cities in China")
plt.ylabel("Population")
plt.show()
```

This chart is a simple visualisation of the population of the urban areas or cities. I want to note that the data about population in chinese cities or areas can easily vary depending on what is counted as a certain region in that database. Population for a place with the same name can vary a lot， for example if you search Chongqing population, some data shows the whole region around the urban area too, aka the county basically.

```python
import seaborn as sns
sns.boxplot(data=df, x="Population Estimate (2025)", color="pink")
```

This boxplot data shows very well how some cities in China have a much bigger population compared to the rest. My data contains 30 urban areas, and on the boxplot chart the bubbles represent the 4 cities that have a noticeably bigger population compared to the rest of the data.

```python
plt.scatter(df["Population Estimate (2025)"], df["Land Area (Km²)"], s=df["Population Estimate (2025)"]*0.00003, alpha=0.4, color="black")
plt.xlabel("2025 population estimate")
plt.ylabel("Land area (Km²)")
plt.title("Land area by population estimate (2025) bubble size=population")
plt.show()
```

This chart visualises that for most of the cities, the bigger the area, the bigger the population. In the cluster of smaller populations, theres a bunch of slight differences. Between the 0.5 and 1.0 1e7 on the population line, there's a bunch of urban areas in a vertical line, meaning that these have about the same population but on varying amounts of land. Of the two biggest, Shanghai has a bigger population but on quite a lot smaller of an area.

```python
plt.scatter(df["Population Estimate (2025)"], df["Density (P/Km²)"], s=df["Density (P/Km²)"]*0.08, alpha=0.6, color="blue")
plt.xlabel("Population Estimate (2025)")
plt.ylabel("Density (P/Km²)")
plt.title("Density by population estimate (2025) (bubble=density)")
plt.show()
```

After seeing the last chart comparing population to land, I wanted to see how the chart would look like when I replace land area with density. I like that on this chart, the urban areas that were all in a small cluster on the last chart, are now spread apart and it's much easier to understand how big of a difference the slight differences between land area and population make.

```python
df.plot(kind="bar", x="Urban Area", y="Projections (2050)", legend=False, color="pink")
plt.title("2050 projection of the population of urban areas in China")
plt.ylabel("Population projection 2050")
plt.show()
```

I decided to do a chart on the projections for 2050 of the population of these urban areas to see, how big of changes are predicted. Since the cities are listed on the chart in the order of their population in 2025, it's possible to see that even though the first 4 urban areas are in the same order, there's a prediction that some cities will surpass others by population.

I think it's very interesting to see how different charts can give you varying understandings of the same data.
