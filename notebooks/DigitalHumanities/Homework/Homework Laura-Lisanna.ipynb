```python
import pandas as pd
df = pd.read_csv("China cities population.csv")
df
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Urban Area", y="Population Estimate (2025)", legend=False, color="pink")
plt.title("Population of cities in China")
plt.ylabel("Population")
plt.show()
```

```python
import seaborn as sns
sns.boxplot(data=df, x="Population Estimate (2025)", color="pink")
```

```python
plt.scatter(df["Population Estimate (2025)"], df["Land Area (Km²)"], s=df["Population Estimate (2025)"]*0.00003, alpha=0.3, color="pink")
plt.xlabel("2025 population estimate")
plt.ylabel("Land area (Km²)")
plt.title("Land area by population estimate (2025) bubble size=population")
```
